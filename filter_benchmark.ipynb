### Steric filter benchmark

In this interactive notebook, the results of the steric filter benchmark are summarized from `filter_benchmark.ipynb`. These steric rules are derived from a benchmark of theoretical structures. In this benchmark, each substructure class was iterated over with an increasing number of tert-butyl groups as the substituents around the rotatable axis (starting from zero, i.e. all hydrogens) and the rotational barrier computed using the atropisomers workflow with the methods and levels of theory described in the manuscript. The rotational barriers computed by our workflow were then used to infer the maximum allowed number of non-heavy substituents per substructure, using 12 kcal/mol as a conservative threshold. Note that some entries are skipped, but were not needed to define the rules.

#### Imports

```python
import pandas as pd
from rdkit.Chem import AllChem as Chem
from rdkit.Chem import Draw
```

#### Results

```python
# Load the .csv file as a dataframe
df = pd.read_csv('filter_benchmark.csv')
```

```python
# Iterate over each row of the dataframe
for i, (index, row) in enumerate(df.iterrows()):

    # Display only the first 30 entries
    if i >= 30:
        continue

    # Create an RDKit molecule from the SMILES string
    mol = Chem.MolFromSmiles(row['SMILES'])

    # Print the results
    print(f"{row['Index']}. SMILES: {row['SMILES']}")
    print(f"Computed barrier: {row['Computed barrier (kcal/mol)']} (class {row['Computed class']})")

    # Determine if the barrier meets the threshold for scanning
    if row['Computed barrier (kcal/mol)'] > 12:
        print("Barrier meets scan threshold\n")
    else:
        print("Barrier fails scan threshold\n")

    # Display the molecule
    display(Draw.MolToImage(mol))
```
