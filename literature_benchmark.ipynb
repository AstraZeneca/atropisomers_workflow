### Literature benchmark

In this interactive notebook, the results of the literature benchmark are summarized from `literature_benchmark.ipynb`. In this benchmark, the rotational barrier of each structure was computed using the atropisomers workflow with the methods and levels of theory described in the manuscript. The rotational barriers computed by our workflow were then compared to the literature values.
#### Imports

```python
import pandas as pd
from rdkit.Chem import AllChem as Chem
from rdkit.Chem import Draw
```

#### Results

```python
# Load the .csv file as a dataframe
df = pd.read_csv('literature_benchmark.csv')
```

```python
# Iterate over each row of the dataframe
for i, (index, row) in enumerate(df.iterrows()):

    # Display only the first 50 entries
    if i >= 50:
        continue

    # Create an RDKit molecule from the SMILES string
    mol = Chem.MolFromSmiles(row['SMILES'])
    
    # Display the molecule
    display(Draw.MolToImage(mol))

    # Print the results
    print(f"{row['Index']}. SMILES: {row['SMILES']}")
    print(f"Substructure: {row['Substructure']}")
    print(f"Workflow computed in: {row['Computation solvent']}")
    print(f"Computed barrier: {row['Computed barrier (kcal/mol)']} (class {row['Computed class']})")
    print(f"Literature barrier: {row['Literature barrier (kcal/mol)']} (class {row['Literature class']})")
    print(f"Source: {row['Literature source']}\n")
```
