### Substructure demonstration

In this interactive notebook, the breadth of substructures the atropisomers workflow can be applied to is demonstrated using the cheminformatics portion of the atropisomers workflow.

#### Imports

```python
# Import the atropisomers class
from atropisomers.run import Atropisomers
```

#### Substructure examples

```python
def parse(smiles):
    """Helper function to run atropisomer detection cheminformatics."""
    print(f"\n### Parsing SMILES: {smiles}\n")

    # Initialize the Atropisomers class with the SMILES string and dummy variables
    m = Atropisomers(
        job_name="example",
        job_dir="./example",
        tmp_dir="./example",
        smiles=smiles,
        bashrc="",
        xtb="",
        crest="",
        elim="",
        censo="",
        censorc="",
        cosmotherm="",
        gaussian="",
        )

    # Run the cheminformatics analysis with steric filters enabled and coupled benzamide analysis disabled
    m.cheminformatics(filter=True, coupled_benzamide=False)
```
