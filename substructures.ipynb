### Substructure demonstration

In this interactive notebook, the breadth of substructures the atropisomers workflow can be applied to is demonstrated.

#### Imports

```python
# General imports
import subprocess
from pathlib import Path

# Import the atropisomers class
from atropisomers import Atropisomers
```

#### Substructure examples

```python
def parse(smiles):
    """Helper function to run atropisomer detection."""
    print(f"### Parsing SMILES: {smiles}\n")
    m = Atropisomers(smiles)
    m.cheminformatics(filter=True, coupled_benzamide=False)
```

General substructure coverage:

```python
# Biaryls
parse('Cc1cccc(C)c1c2c(C)cccc2Cl') # 6,6
parse('Cc1occ(C)c1c2c(C)cccc2C') # 6,5
parse('Cc1occ(C)c1c2cocc2C') # 5,5
parse("CCOC(=O)c1c(C)[nH]c(c2ccccc2)c1c3c([nH]c(C)c3C(=O)OCC)c4ccccc4") # Literature 5,5
parse("CC(=O)N[C@H]1CCC2=CC(=C(C(=C2C3=CC=C(C(=O)C=C13)OC)OC)OC)OC") # Colchicine (7-membered, fused ring)

# Benzamides and amides
parse('NC(=O)c1cccc2CCCC(=O)c12') # Benzamide
parse('NC(=S)c1cccc2CCCC(=O)c12') # Thiobenzamide
parse('CC(=O)Nc1c(C)ccc2CCCC(=O)c12') # Reversed benzamide
parse("CC1CCCC(=O)N1c2c(C)cccc2C") # Cyclic benzamide
parse("Cc1ccoc1C(=O)N") # 5-membered benzamide
parse("CC(=O)N") # Acetamide
parse("CN(C)C(C)=S") # Thioamide

# Sulfones and sulfoxides
parse("CS(=O)c1c(C)ccc2CCCC(=O)c12") # 6-membered sulfoxide
parse("CS(=O)c1c(C)oc2CCCC(=O)c12") # 5-membered sulfoxide
parse("CS(=O)(=O)c1c(C)ccc2CCCC(=O)c12") # Sulfone

# Sulfinamide and sulfonamide
parse("CN(c1c(C)cccc1C)S(=O)C") # Sulfinamide
parse("CN(c1c(C)cccc1C)S(=O)(=O)C") # Sulfonamide

# Diaryls
parse("Cc1ccccc1Oc2c(C)cccc2C") # Ether
parse("Cc1ccccc1Nc2c(C)cccc2C") # Amine
parse("Cc1ccccc1Sc2c(C)cccc2C") # # Thioether
parse("c1cc(cc(c1S(=O)c2ccc(cc2C)C)C)C") # Sulfoxide
parse("c1cc(cc(c1S(=O)(=O)c2ccc(cc2C)C)C)C") # Sulfone

# Aryl-aliphatics
parse("CC(C)(C)c1cc(N2CCCCC2)c(C(C)(C)C)cc1Cl") # 6,6
parse("Oc1c(c(cc(c1)C)O)C2[C@@H](CCC(=C2)C)C(=C)C") # Cannabidiol (6,6)
parse("CC1CCC(C)C1c2c(C)cccc2C") # 6,5
parse("CC1CCCC(C)C1c2c(C)occ2C") # 5,6
parse("CC1CCC(C)C1c2c(C)occ2C") # 5,5

# Bialiphatics
parse("CC1=C(C(=CC1)C)C2=C(C)CC=C2C") # sp2 rotatable bond
parse("CC1=CC=C(C)C1C2C(=CC=C2C)C") # sp3 rotatable bond
```

Fused rings are only allowed if the fused ring is 7-membered or more:

```python
# Biaryl
parse("C1Cc2ccccc2c3ccccc13")
parse("C1CCc2ccccc2c3ccccc13")

# Diaryls
parse("C1c2ccccc2Oc3ccccc13")
parse("C1Cc2ccccc2Oc3ccccc13")

# Aliphatics
parse("C1CCC2C(C1)CC3CCCCC23")
parse("C1CCC2C(C1)CCCCC3CCCCC23")

# Benzamides
parse("O=C1NCCc2ccccc12")
parse("O=C1NCCCc2ccccc12")

# Amides
parse("O=C1CCCCN1")
parse("O=C1CCCCCCCN1")
```
